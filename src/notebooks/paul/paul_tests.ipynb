```python
import pandas as pd

# read the data from ../../data/final_data.csv
data = pd.read_csv('../../../data/final_data.csv', engine='python')
data.head()
```

```python
# print the number of total rows
total_num_of_rows = data.shape[0]
# count the number of rows that have more than 1 column with value -1
number_of_rows_with_more_than_1_missing_value = data[(data == -1).sum(axis=1) > 1].shape[0]

# print the percentage of rows that have more than 1 column with value -1
print('Percentage of rows that have more than 1 column with missing values: ',
      number_of_rows_with_more_than_1_missing_value / total_num_of_rows * 100)

# number of rows with exactly 1 missing value
number_of_rows_with_1_missing_value = data[(data == -1).sum(axis=1) == 1].shape[0]

# print the percentage of rows that have exactly 1 column with value -1
print('Percentage of rows that have exactly 1 column with missing values: ',
      number_of_rows_with_1_missing_value / total_num_of_rows * 100)
```
